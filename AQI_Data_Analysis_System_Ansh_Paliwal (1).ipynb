## 1. Import libraries

```python
import pandas as pd
import numpy as np
```

## 2. Create the sample dataset

```python
data = {
    "City": ["Delhi", "Delhi", "Mumbai", "Mumbai", "Lucknow", "Lucknow",
             "Chennai", "Chennai", "Kolkata", "Kolkata"],
    "Date": ["2026-01-01", "2026-01-02", "2026-01-01", "2026-01-02",
             "2026-01-01", "2026-01-02", "2026-01-01", "2026-01-02",
             "2026-01-01", "2026-01-02"],
    "AQI": [280, 310, 125, 140, 190, 175, 75, 82, 160, 170],
    "PM2.5": [150, 170, 65, 72, 95, 88, 35, 40, 80, 85],
    "PM10": [220, 240, 110, 120, 160, 150, 70, 75, 135, 145]
}
df = pd.DataFrame(data)
df["Date"] = pd.to_datetime(df["Date"])
print(df)
```

## 3. Inspect the dataset

```python
print("First five rows:")
print(df.head())
print("\nRows and columns:", df.shape)
print("\nData types:")
print(df.dtypes)
```

## 4. Check and clean the dataset

```python
print("Missing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()
df = df.dropna(subset=["AQI"])
print("\nCleaned dataset:")
print(df)
```

## 5. Calculate AQI statistics using NumPy

```python
aqi_values = df["AQI"].to_numpy()
print("Average AQI:", round(np.mean(aqi_values), 2))
print("Maximum AQI:", np.max(aqi_values))
print("Minimum AQI:", np.min(aqi_values))
print("Standard deviation:", round(np.std(aqi_values), 2))
```

## 6. Calculate city-wise average AQI using Pandas

```python
city_average = df.groupby("City")["AQI"].mean().sort_values(ascending=False)
print(city_average)
print("\nHighest average AQI city:", city_average.idxmax())
print("Lowest average AQI city:", city_average.idxmin())
```

## 7. Classify AQI readings

```python
def classify_aqi(value):
    if value <= 50:
        return "Good"
    elif value <= 100:
        return "Satisfactory"
    elif value <= 200:
        return "Moderate"
    elif value <= 300:
        return "Poor"
    elif value <= 400:
        return "Very Poor"
    else:
        return "Severe"

df["AQI_Category"] = df["AQI"].apply(classify_aqi)
print(df[["City", "Date", "AQI", "AQI_Category"]])
```

## 8. Find the record with the highest AQI

```python
highest_record = df.loc[df["AQI"].idxmax()]
print(highest_record)
```

## 9. Count AQI categories

```python
print(df["AQI_Category"].value_counts())
```

## 10. Save the result as CSV

```python
df.to_csv("aqi_analysis_result.csv", index=False)
print("Saved: aqi_analysis_result.csv")
```
